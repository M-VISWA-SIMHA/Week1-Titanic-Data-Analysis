- Pandas – for data manipulation and analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv("titanic_cleaned.csv")
```

## 7. Visualization 1 – Overall Passenger Survival

### Objective

The purpose of this visualization is to compare the number of passengers who survived with those who did not survive.

### Chart Type

A count plot is used because it provides a simple comparison of the two survival categories.

### Interpretation

The visualization shows that the number of passengers who did not survive was higher than the number of passengers who survived.

This provides an overall understanding of the survival outcome before examining individual factors.

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df, x="Survived")

plt.title("Titanic Passenger Survival")
plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")

plt.show()
```

## 8. Visualization 2 – Survival by Gender

### Objective

The purpose of this visualization is to examine whether survival outcomes differed between male and female passengers.

### Chart Type

A grouped count plot is used to compare survivors and non-survivors for each gender.

### Interpretation

The visualization shows a clear difference in survival outcomes between male and female passengers.

Female passengers have a substantially higher number of survivors relative to their non-survivors, while male passengers have a much larger number of non-survivors.

### Key Insight

Gender appears to be strongly associated with survival outcomes in the Titanic dataset.

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="Sex",
    hue="Survived"
)

plt.title("Titanic Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

plt.show()
```

## 9. Visualization 3 – Survival by Passenger Class

### Objective

This visualization examines the relationship between passenger class and survival.

### Chart Type

A count plot is used to compare survival outcomes across the three passenger classes.

### Interpretation

The visualization allows us to compare the number of survivors and non-survivors in First, Second, and Third Class.

### Key Insight

Passenger class appears to be an important factor associated with survival. Survival outcomes differ across passenger classes.

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="Pclass",
    hue="Survived"
)

plt.title("Titanic Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.show()
```

## 10. Visualization 4 – Age Distribution by Survival

### Objective

This visualization examines the age distribution of passengers according to their survival status.

### Chart Type

An age distribution plot is used to compare the age patterns of survivors and non-survivors.

### Interpretation

The visualization helps identify whether certain age groups appear more frequently among survivors or non-survivors.

### Key Insight

Age provides additional information about the characteristics of passengers associated with different survival outcomes.

```python
plt.figure(figsize=(10,6))

sns.histplot(
    data=df,
    x="Age",
    hue="Survived",
    bins=30,
    kde=True
)

plt.title("Age Distribution by Survival Status")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()
```

## 11. Visualization 5 – Survival Rate by Passenger Class and Gender

### Objective

This visualization examines survival by considering two factors together: passenger class and gender.

### Chart Type

A grouped visualization is used to compare survival outcomes across passenger classes and gender.

### Interpretation

Combining passenger class and gender provides a more detailed view than examining either factor separately.

### Key Insight

The visualization suggests that survival was associated with both gender and passenger class, and that these factors should be considered together when interpreting Titanic survival patterns.

```python
plt.figure(figsize=(10,6))

sns.barplot(
    data=df,
    x="Pclass",
    y="Survived",
    hue="Sex"
)

plt.title("Survival Rate by Passenger Class and Gender")
plt.xlabel("Passenger Class")
plt.ylabel("Average Survival Rate")

plt.show()
```

## 12. Visualization 6 – Correlation Heatmap

### Objective

The correlation heatmap is used to examine relationships between numerical variables in the Titanic dataset.

### Chart Type

A heatmap is used because it provides an easy visual representation of correlation values between variables.

### Interpretation

The heatmap helps identify which numerical variables have stronger or weaker relationships with survival and with one another.

### Key Insight

Correlation analysis provides another perspective on the relationships within the dataset, although correlation alone does not prove causation.

```python
numeric_columns = [
    "Survived",
    "Pclass",
    "Age",
    "SibSp",
    "Parch",
    "Fare"
]

correlation = df[numeric_columns].corr()

plt.figure(figsize=(9,7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Between Titanic Variables")

plt.show()
```

## 13. Interactive Visualization

### Objective

An interactive Plotly visualization is created to allow users to explore the Titanic dataset more dynamically.

### Why Interactive Visualization?

Interactive charts allow users to hover over data points, inspect values, and explore relationships more easily than static charts.

### Interpretation

The interactive visualization provides an additional way to explore passenger characteristics and survival outcomes.

It improves the presentation of the analysis by allowing the user to investigate the data interactively.

```python
fig = px.scatter(
    df,
    x="Age",
    y="Fare",
    color="Survived",
    size="Pclass",
    hover_data=["Sex", "Pclass"],
    title="Titanic Passengers: Age, Fare and Survival"
)

fig.show()
```
